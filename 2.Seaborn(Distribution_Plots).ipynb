We use distribution plots to visualize the distribution of quantitative data.

Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

1. distplot/histplot (imp)
2. jointplot  (imp)
3. pairplot (imp)
4. rugplot

```python
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')           # Load the dataset , dataset name -- tips
#  Here see the dataset and choose what is quantative and catagorical
# here Sex is catagorical values so we  can not do any operations in here as it distribution chapter
df
```

```python
df['size'].unique()   # Here only 6 unique values out of 244  rows , so now total_bill  & tips are quantative
```

```python
                                   #  1. distplot/histplot
```

```python
sns.displot(df['total_bill'])    # SNS is  use to make good type of plot    # To see distrbutions it will help more(like the example)
```

```python
sns.displot(df['total_bill'],bins = 80)   # we can show , how much bins i need
                                               # bins = 80 here
```

```python
# Kernal Density Estimator(kde)   # To show dept 
sns.displot(df['total_bill'],bins = 30,kde = True)
```

```python
plt.subplot(1,2,1)
sns.displot(df['total_bill'],bins = 30,kde = True)    # To plot two data together
plt.subplot(1,2,2)
sns.displot(df['tip'],bins = 30,kde = True)
```

```python
                                  #2. Joint Plot( EG - comparing two column like here we will compare Total_bill & tips)
```

```python
sns.jointplot(x = "total_bill",y = 'tip',data = df,kind = 'scatter')  # x axis , y axis , data = data location where data store, 
                                                                           #scatter --  
```

```python
          # Scatter point note from you tibe ss
from IPython.display import Image

# Just provide the exact file name and desired width
Image(filename="Scatter_point [seaborn joint] (make a dot)(By use we make linear regression) imp.png", width=400)
     # Here dot is scatter point and using scatter point we make lineat regression later
```

```python
sns.jointplot(x = "total_bill",y = 'tip',data = df,kind = 'kde') # using kde kind 
```

```python
sns.jointplot(x = "total_bill",y = 'tip',data = df,kind = 'hex')  # using hex kind
```

```python
sns.jointplot(x = "total_bill",y = 'tip',data = df,kind = 'reg')  # using reg kind
```

```python
                # 3. pairplot (imp)
```

```python
sns.pairplot(df)         # just pair all and comparison with others
```

```python
sns.pairplot(df,hue = 'sex')       # Hue mean we divide data by catagoraical ( eg- male or female which catagory gives more bill)
```

```python
sns.pairplot(df,hue = 'size',palette = 'rainbow')  # palette -- for colour
```
