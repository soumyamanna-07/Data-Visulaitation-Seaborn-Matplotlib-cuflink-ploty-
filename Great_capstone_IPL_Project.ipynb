## 1.Loading the libraries and dataset

## 4.Now here come some Basic Questions

```python
match_wins = df['match_winner'].value_counts()        # Values_counts() make a series 
sns.barplot(y = match_wins.index , x = match_wins.values,palette='viridis')     #make a bar plot
plt.title("Most Match Win By Team")
```

```python
sns.countplot(x = df['toss_decision'],palette='rainbow')
plt.title("Toss Decision Trends")
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count() # Want to make percentage
percentage = (count *100)/df.shape[0]
print(f"Toss Win and Match Win  percentage is : {percentage.round(2)}")
```

```python
sns.countplot(x = df['won_by'])
plt.title("Won By")
```

## 5.Key Player Pkerformances

```python
count = df['player_of_the_match'].value_counts().head(10)    # Head for top 10
count
```

```python
sns.barplot(x =count.values , y = count.index,palette = 'mako')
plt.title("Top 10 players Man of the match award")
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending = False).head(2)  # Here we need to work by 2 columns so when we need to work 2 or more columns we use group by 
high                                                                                 # sort value assce.. is for top score
                                                                                   # head is for top 2
```

```python
high.plot(kind = 'barh') # h for to see horizontal else you can write bar 
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])                 #Group by data by best bowling then extract best bowling figure
                                                                      # SPLIT  by --  and then find the 0 element vales means wickets
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending = False).head(10)
top_bowlers.plot(kind = 'barh')
```
