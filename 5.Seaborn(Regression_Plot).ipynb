```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")            # Just load dataset
tips
```

```python
sns.lmplot(x = 'total_bill',y = 'tip',data = tips)                                       # Usally In Y axis -- predict value , X axis -- predict thorugh which values
```

```python
sns.lmplot(x = 'total_bill',y = 'tip',data = tips,hue = 'sex')  # using hue for identify male and female                                       # Usally In Y axis -- predict value , X axis -- predict thorugh which values
```

```python
sns.lmplot(x = 'total_bill',y = 'tip',data = tips,hue = 'sex',palette = 'coolwarm')     # You can change colour using palette
```

```python
sns.lmplot(x = 'total_bill',y = 'tip',data = tips,hue = 'sex',palette = 'coolwarm',markers =['o','v'] )                 # you can change marker  
```

```python
sns.lmplot(x = 'total_bill',y = 'tip',data = tips,hue = 'sex',palette = 'coolwarm',markers =['o','v'],scatter_kws = {'s':200}  )               # You can change size also using scatter_kws
```
