```python
                           # 3. Object Oriented way( By that it can acces in canvas and we can change in canvas)
```

```python
fig  = plt.figure()  # 1st we create a object ( It will not give any output, it just a canvas)
```

```python
fig = plt.figure()                 ## The inside value must be  between 0 and 1
axis1 = fig.add_axes([0.1,0.2,0.7,0.8])     # o.1 -- how many percentage you need to come from left          
                                     #0.2 -- How many percentage you need to come from Button
                                      #0.7 -- It means that 70% your width
                                        # 0.8 -- It means that 80% your hight of the canvas
```

```python
fig = plt.figure()                
axis1 = fig.add_axes([0.1,0.2,0.7,0.8])     # create a axis
axis1.plot(x,y)         # plot in axis1
```

```python
fig = plt.figure()                
axis1 = fig.add_axes([0.1,0.2,0.7,0.8])     # create a axis
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.5])  # Pass to axis 
axis2.plot(x,y)              # by adjust 0.3 , 0.4 you can adjust location of those 2nd canvas 
```

```python
 # we can adjust size of canvas also let's do it
fig = plt.figure(figsize = (10,11))     # it's for resize of canvas -- figsize = (10,11)  . 10 -- Row size , 11 -- Column size         
axis1 = fig.add_axes([0.1,0.2,0.7,0.8])     # create a axis
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.5])  # Pass to axis 
axis2.plot(x,y)
```

```python
 # We can image or canvas quality by gain DPI ( dots per inch)
fig = plt.figure(figsize = (10,11),dpi = 800)   # selectk dpi = 800   # Auto DPI selct is 100     
axis1 = fig.add_axes([0.1,0.2,0.7,0.8])     # create a axis
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.5])  # Pass to axis 
axis2.plot(x,y)
```
